```python
import json
import os

# 원본/저장 폴더 설정
src_folder = r"C:/00AI/LLM/21project3/2025-W45"
dst_folder = r"C:/00AI/LLM/21project3/2025-W45-1"
os.makedirs(dst_folder, exist_ok=True)

def replace_key(obj):
    """JSON 전체 구조에서 content → context 변환"""
    if isinstance(obj, dict):
        new_obj = {}
        for k, v in obj.items():
            new_key = "context" if k == "content" else k
            new_obj[new_key] = replace_key(v)
        return new_obj
    elif isinstance(obj, list):
        return [replace_key(item) for item in obj]
    else:
        return obj

# 폴더 내 JSON 파일 반복 처리
for filename in os.listdir(src_folder):
    if filename.endswith(".json"):
        src_path = os.path.join(src_folder, filename)
        dst_path = os.path.join(dst_folder, filename)  # 원본과 동일한 파일명 사용

        try:
            with open(src_path, "r", encoding="utf-8") as f:
                data = json.load(f)

            # content → context 변환
            converted = replace_key(data)

            # 새 폴더로 저장 (파일명 그대로)
            with open(dst_path, "w", encoding="utf-8") as f:
                json.dump(converted, f, indent=2, ensure_ascii=False)

            print(f"🔄 변환 완료: {filename}")

        except Exception as e:
            print(f"❌ 변환 실패: {filename} ({e})")
```
