```python
# TypeDict
from typing import TypedDict

class User(TypedDict):  # --> dict처럼 사용
    name: str
    age : int

u : User = {
    'name': '홍길동',
    'age' : 25
}

print(type(u))
print(u)
print(u['name'])

u : User = {
    'name': 100, # 상기 str은 가이드 일뿐 무조건 강제는 아님
    'age' : 25
}

print(type(u))
print(u)
print(u['name'])
```

```python
from pydantic import BaseModel
class User(BaseModel):
    name: str
    age : int

user = User(name='강감찬', age=20)
print(type(user))
print(user)


user = User(name=500, age=20)  # 상기 함수는 강제성이 있기 때문에 name에 str이 안오면 에러남!!!
print(type(user))
print(user)
```
