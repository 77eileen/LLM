````python
# 자동 미분 (autograd) : 매 계산과정마다 계산과정의 미분값을 저장
# 파이토치 required_grad = True로 설정하면 : 계산그래프라는 형태로 기록 - 동영상녹화와 비슷한 원리
# .backword() 호출하면 이 과정을 거꾸로 되감으면서 각 단계의 미분값을 계산해 최종 기울기를 얻는다 
````

````python
x = torch.ones(2,2, requires_grad=True)
x, x.size()
````

````python
y = x + 2  #y도 계산그래프에 포함이됨 (x에서 requires_grad=True 이렇게 되어 있으므로)
print (y, y.grad_fn)  
z= y*y*3    #x를 계산그래프에 기록되어 있으므로, 연쇄적으로 z, out 도 모두 계산그래프에 기록됨.
out = z.mean()
z, out
````

````python
# 딥러닝에서는 기울기를 구하는 것이 가장 중요함.
# 기울기를 구하는 과정을 역전파(backword <-> forward)라 함 -> 기울기 계산

# 최종 결과가 out 에서부터 시작해서 계산그래프를 거꾸로 거슬러 올라가면서 미분의 연쇄법칙을 사용해서 미분값을 계산
# out을 x에 대해서 미분
out.backward()
x.grad
````

````python
# out = 1/4(z1+z2+z3+z4)
# z = 3y^2
# y = x + 2

# d-out/dxi =
# chat gpt를 이용해서 미분 연쇄법칙으로 계산해보면 상기 x.grad 와 동일한 4.5 결과값이 나옴
````

````python
# 기울기 제어
# 역전파, torch.no_grad, detach
# required_grad =True 연산 추적
import torch
x = torch.randn(3, requires_grad=True)
print(x)
# y는 x로부터 연산기록을 물려받음
y = x*2
y, y.data, y.data.norm() 
# y.data: 텐서의 기록을 무시하고 순수한 데이터만 출력
# norm(노름, 유클리드 노름) : 벡터의 크기를 측정하는 방법 (원점으로부터 벡터까지의 거리) 모든 요소를 각각 제곱하고 모두 더해서 결과에 제곱근(루트)
# y의 실제데이터의 크기를 확인해서 반복 
i=0
while y.data.norm() < 1000:
    y=y*2
    i += 1
print (f'y: {y}')
print (f'반복횟수 : {i} y=x*2^i+1')

# 벡터에 대한 역전파 ( 야코비안 행렬 )
# 벡터를 벡터로 미분하면 결과는 행렬인 야코비안 행렬이라 일컫음.
# pytorch는 전체 야코비안 행렬을 직접 계산하는 대신 야코비안-벡터 곱을 계산해서 메모리를 절약
v=torch.tensor([ 0.1, 1.0, 0.0001], dtype=torch.float)
y.backward(v)  #y를 x에 대해 미분한 야코비안 행렬과 입력으로 제공된 벡터 v를 곱하라는 의미
print(f'x.grad : {x.grad}')
````

💡 1️⃣ “기울기(Gradient)”가 왜 중요한가?

기울기는 “변화율”, 즉 **“얼마나, 어느 방향으로 바꿔야 더 좋아질지”**를 알려줍니다.

예를 들어볼게요 👇

모델이 예측을 했는데,
“예측값이 정답보다 조금 작다”

이때 우리는
“그럼 가중치를 얼마만큼 늘려야 예측이 맞아질까?”
이걸 알고 싶어요.

👉 **그게 바로 기울기(gradient)**예요.

🧭 2️⃣ “야코비안 행렬”은 어디에 쓰는가?

기울기가 스칼라 → 벡터 관계였다면,
야코비안은 벡터 → 벡터 관계에서의 “기울기”입니다.

```
▶ 스칼라 백워드 (Scalar Backward) : 역전파로 구하는 기울기가 하나일때
out.backwqrd()
최종점수가 숫자하나.  ex) 수학공부 1시간 했을 때 최종점수에 몇점의 영향을 줬는지 
최종목표가 하나 (스칼라)이기 때문에 각 원인(각 변수)의 기여도를 바로 계산 

▶ 벡터 백워드 vector backward : 기울기가 여러개 일 때
y.backward (v)
ex. 과목별(국영수) 점수가 담긴 성적표 (벡터)가 있을 때, 
공부시간이 성적에 얼마나 영향을 줬는지 물어보면 애매함. 국어성적이 오르나? 영어성적이 오르나?
가중치 또는 중요도를 알려주는 벡터 v=torch.tensor([ 0.1, 1.0, 0.0001], dtype=torch.float)
이 중요도를 반영해서 최종적으로 공부시간의 기여도를 확인 
```

````python
x.requires_grad
# 각 학습을 진행하면서 도중에 모델의 성능 평가가 필요 - 기울기 계산을 하면 안됨
with torch.no_grad():   # no_grad 는 기울기를 계산하지 않고, 선택값을 출력?
    print(f'no_grad 내부 : {(x**2).requires_grad}')

# .detach() 계산그래프 
print(f'detach 전 {y.requires_grad}')
y=x.detach()
print(f'detach 후 {y.requires_grad}')
````
