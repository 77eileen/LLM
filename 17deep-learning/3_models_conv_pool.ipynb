##### AlexNet

[입력 이미지]
   ↓
합성곱 → 풀링 → 합성곱 → 풀링
   ↓
완전연결층 → 완전연결층 → 결과(예: 고양이)

1. 합성곱층 (Convolution Layer)
- 이미지를 조금씩 살펴보면서 특징을 뽑는 단계야.
- 예를 들어 고양이 사진이라면,
<br>   처음엔 “귀 모양”이나 “눈동자” 같은 패턴을 찾아내는 식이야.
<br>   이 과정을 여러 번 반복하면서
<br>   “아~ 이건 고양이 얼굴이구나!” 하고 점점 배워가는 거야.
- 비유: 사진을 확대경으로 부분부분 자세히 살펴보는 단계

2. 풀링층 (Pooling Layer)
- 너무 세세한 정보는 줄이고, 핵심만 남기는 요약 단계야.
- 예를 들어 사진의 위치가 살짝 달라져도 결과가 크게 변하지 않게 만들어줘.
<br> (즉, 더 똑똑하고 안정적인 모델이 되게 함!)
- 비유: 큰 그림을 보기 위해 디테일은 살짝 흐리게 만드는 느낌!

3. 완전연결층 (fullyConnected Layer, FC Layer)
- 앞에서 찾은 특징들을 종합해서 “이건 고양이!” “이건 자동차!” 처럼 최종 판단을 내리는 단계야.
- 비유: 앞의 단서들을 종합해서 결론을 내리는 “심사위원” 같은 역할!

- 🍪 쿠키 공장 비유
- 당신이 쿠키 60,000개를 구워야 해요.
- 오븐(=GPU)은 한 번에 128개밖에 못 구워요.
- 그래서 한 번에 128개씩 구워서 여러 번 반복합니다.
<br> 👉 이때 “한 판에 128개”가 바로 배치(batch)
<br> 👉 “60,000개 전체”는 전체 데이터셋(dataset)
<br> 👉 “한 판 다 굽고 다음 판으로 넘어가는 횟수”는 스텝(step)
<br> 👉 “모든 쿠키 60,000개를 다 구운 후 다시 반복 시작하는 것”은 에포크(epoch)

| 개념                     | 의미                    | 쿠키 공장 비유                    |
| ---------------------- | --------------------- | --------------------------- |
| **Dataset (데이터셋)**     | 전체 학습 데이터             | 쿠키 반죽 전체 (예: 60,000개 쿠키 반죽) |
| **Batch (배치)**         | 한 번에 모델이 처리하는 데이터 묶음  | 🍪 **오븐 1판**                |
| **Batch Size (배치 크기)** | 그 묶음(판)에 들어 있는 데이터 개수 | 한 판에 올린 쿠키 **개수 (예: 128개)** |
| **Step (스텝)**          | 배치 1개를 처리하는 과정        | 오븐에서 쿠키 1판 굽기               |
| **Epoch (에포크)**        | 모든 데이터를 한 번 다 학습한 것   | 반죽 60,000개 다 구워서 한 바퀴 끝내기   |

- 텐서 : 숫자 담는 상자 / 숫자가 어떤 구조로 모여있느냐. (스텔라,벡터,행렬, 흑백,배치단위이미지) (넘파이 배열의 딥러닝)



```python
import torch
import torch.nn as nn
```

```python
# 배치크기(샘플의 갯수) x 채널크기 x 높이 x 너비   
inputs = torch.Tensor(1,1,28,28)  # 28x28크기를 갖는 한개의 흑백이미지
print(f'텐서의 크기 : {inputs.size()}')
```

```python
# 합성곱층과 폴링을 선언
conv1= nn.Conv2d(1,32,3, padding=1) 
# 합성곱 nn.Conv2d(in_ch, out_ch, kernel_size, padding=...)
# 입력채널 1 (위에 inputs가 1이라서) / 출력채널 32 (1x28x28 입력이 conv2d 통과하면 32x28x28 로 출력) / 커널 3 / 패딩1
print(conv1)
```

```python
# 2번째 합성곱 층
# tensorflow는 레이어를 정의할때 출력할 뉴런의갯수만 정해주면 되는데, torch는 다 알려줘야함
conv2=nn.Conv2d(32,64,3,padding=1)
# 32는 위에 conv1에서 출력채널이 32였기 때문에 input채널로 32로 줌
print(conv2)
# 층마다 어떻게 조립하나 --> 하이퍼파라미터 (우리가 만들기에 따라서 성능 차이남)
```

```python
# 풀링층
pool = nn.MaxPool2d(2) #maxPooling
print(pool)

# stride =2 : 2칸씩 이동?
# 4x4는 풀링층을 지나면 2x2 가 됨 ==> 특성 맵 나누기 stride (4x4)/(2x2)=2
```

```python
# 데이터 통과하기
# inputs.size()  #torch.Size([1, 1, 28, 28])
out = conv1(inputs)
out.size()            #상기 conv1 에서 예상한 결과로 나옴... conv1에 패딩안주면 1,32,26,26이 나옴
```

```python
out2 = pool(out)
print(out2.size())     #1 32 14 14
```

```python
# 풀링한 것을 out하면?
out3 = conv2(out2)
print(out3.size())
```

```python
out4 = pool(out3)
print(out4.size())
```

```python
# 밀집층 (fc)
out5 = out4.view(out.size(0), -1) #1차원으로 펼침
print (out5.size())
```

```python
fc = nn.Linear(3136, 10)   #데이터input은 상기 out5 size 넣고, 출력 10개(임의지정)
out6=fc(out5)
print(out6.size())
```

```python
# 모델 설계

import torch.nn as nn

# nn.Module: PyTorch에서 모든 신경망의 부모클래스

class CNN(nn.Module):
        
        # __init__ 
        # 모델이 생성될때 한번 호출되는 초기화 함수.
        # layer 을 선언
        # 합성곱, 풀링, FC층   
    def __init__ (self):
        super(CNN, self).__init__()  #부모생성자 넣기 모듈넣으면 필수.. # 64 1 28 28


        # 합성곱 nn.Conv2d(in_ch, out_ch, kernel_size, padding=...)
            # in_ch : 입력 채널 수(흑백=1, 컬러=3)
            # out_ch : 필터(커널) 개수 → 출력 채널(특징맵 수). 예: 32면 32개의 서로 다른 필터가 이미지에서 특징을 뽑음.
            # kernel_size=3 : 필터 크기 3×3.
            # padding=1 : 입력 주변에 1픽셀 0(보통)을 채워서 출력 크기가 입력과 같게 유지하도록 함.
            # stride(보폭) : 생략하면 기본값 1. 한 번에 한 칸씩 필터를 이동.
        self.conv1 = nn.Conv2d(1, 32, 3, padding=1)
        self.conv2 = nn.Conv2d(32,64,3,padding=1)

        # 풀링 nn.MaxPool2d(2)
            # 커널이 2×2이고 기본 stride=2 → 가로/세로 크기를 절반으로 줄임.
            # MaxPool은 영역 내 최대값 선택 → 가장 강한 특징만 남김 (노이즈 제거, 위치 변화에 강해짐)
        self.pool = nn.MaxPool2d(2)
        self.fc1 = nn.Linear(64*7*7, 128)  #inpur layer
        self.fc2= nn.Linear(128, 10) #hidden layer 이면서 output layer
    
    
    # forward 함수
    # PyTorch에서 모든 신경망(nn.Module)클래스는 입력데이터를 받아서 출력까지 계산하는 과정을 정의
    # 입력 -> 연산 -> 출력 의 순서를 명시하는 함수
    # CNN층, 풀링, FC층을 거쳐 최종 예측을 출력하는 과정을 적어줌
    def forward(self,x):    #모듈에 있는 forward를 재정의
        x = self.pool(F.relu(self.conv1(x)))   # 최초 (64,1,28,28) --> conv1 출력 (64, 32, 28, 28)  ---> pool 출력 (64,32,14,14)
        x = self.pool(F.relu(self.conv2(x))) # --> conv2 (64,64, 14,14) --> pool 출력 (64, 64, 7, 7) ==> 64*7*7 만 self.fc1에 넣기

        # 중요팁: 배치 크기를 64로 고정해서 x.view(64, -1) 하는 건 좋지 않다.
        # 배치 크기가 바뀌면 에러가 남. 대신 x = x.view(x.size(0), -1) 로 쓰자.
        x = x.view(x.size(0), -1) # x = x.view(64, -1) flatten 
        x = F.relu(self.fc1(x))   # 오류: 입력 행렬 크기 (64×1568) 와 가중치 행렬 (3136×128) 의 크기가 안 맞아서 곱셈 불가능 --> x.view(x.size(0), -1)
        out = self.fc2(x)
        return out
    

model = CNN()
model

# torch.device : PyTorch에서 데이터를 어디서 계산할지 지정하는 코드
# 없으면 cpu사용, 가능하면 GPU(cuda)로 보내 연산속도 향상
# 입력텐서와 레이블도 .to(device)로 이동시켜야 모델과 같은 장치에 있어 오류가 안남
device = torch.device('cuda' if torch.cuda.is_available() else 'cpu')
model.to(device)

# 손실함수 / 다중클래스에 적합 / 내부에 [Log softmax + NLL Loss]가 있음
# 손실함수 : 정답과 예측 사이의 거리를 나타내는 수치
# 모델이 얼마나 틀렸는지 수치로 계산
criterion = nn.CrossEntropyLoss() 

# optimizer 경사하강법(SGD) 변형/ 기울기를 조절
# 가중치를 업데이트(학습)하는 방법을 정함 --> 모델이 틀린 방향을 반영해서 가중치를 수정하는 역할 (오차를 줄이도록)
# 딥러닝 학습은 크게 두 단계로 이루어져요:
# Forward: 모델에게 데이터를 보여주고 점수(score)를 계산하는 과정 
    # 입력 → 모델 → 예측값 계산 → 손실 계산
    # 예: inputs → CNN → outputs → loss
# Backward: 손실값을 기반으로 각 가중치가 얼마나 틀렸는지 계산 
    # 손실 → 역전파 → 가중치 업데이트
    # 실제 데이터와 모델 예측값의 차이(loss)를 줄이도록 가중치 조절
    # PyTorch에서 loss.backward()

# 백워드(backward)와 옵티마이저(optimizer)의 역할 차이
# 1️⃣ loss.backward() → 기울기(gradient) 계산
    # 모델이 틀린 정도(손실)를 바탕으로 각 가중치가 얼마나 잘못됐는지 계산
    # 가중치를 실제로 바꾸지는 않아요!
    # 그냥 **기울기(grad)**를 tensor.grad에 저장해주는 과정이에요.

# 2️⃣ optimizer.step() → 가중치 업데이트
    # 계산된 기울기를 실제로 가중치에 적용
    # learning rate 등 규칙에 맞춰 가중치를 조금씩 조정
    # 이걸 안 하면 기울기는 계산되지만, 모델은 학습되지 않아요.

# 🔹 쉽게 비유하면
# Backward: 지도를 그리는 과정 → 어디로 가야 할지 표시
# Optimizer: 지도를 보고 실제로 발을 움직이는 과정 → 가중치 업데이트

# 그러니까 둘 다 필요합니다.
# loss.backward() 없으면 기울기를 모름 → 어디로 가야할지 모름
# optimizer.step() 없으면 기울기를 계산만 하고 가중치는 그대로 → 학습 안 됨



# optimizer가 바로 역전파로 계산된 기울기를 이용해 가중치를 조정해 주는 도구에요.
optimizer = optim.Adam(model.parameters(), lr=0.001)

```

```python
# 모델클래스 : 상기와 동일

import torch.nn as nn
class CNN(nn.Module):
    def __init__ (self):
        super(CNN, self).__init__() 
        # 합성곱 nn.Conv2d(in_ch, out_ch, kernel_size, padding=...)
            # in_ch : 입력 채널 수(흑백=1, 컬러=3)
            # out_ch : 필터(커널) 개수 → 출력 채널(특징맵 수). 예: 32면 32개의 서로 다른 필터가 이미지에서 특징을 뽑음.
            # kernel_size=3 : 필터 크기 3×3.
            # padding=1 : 입력 주변에 1픽셀 0(보통)을 채워서 출력 크기가 입력과 같게 유지하도록 함.
            # stride(보폭) : 생략하면 기본값 1. 한 번에 한 칸씩 필터를 이동.
        self.conv1 = nn.Conv2d(3, 32, 3, padding=1)
        self.conv2 = nn.Conv2d(32,64,3,padding=1)

        # 풀링 nn.MaxPool2d(2)
            # 커널이 2×2이고 기본 stride=2 → 가로/세로 크기를 절반으로 줄임.
            # MaxPool은 영역 내 최대값 선택 → 가장 강한 특징만 남김 (노이즈 제거, 위치 변화에 강해짐)
        self.pool = nn.MaxPool2d(2)

        # class Linear( in_features: int, out_features: int)
        self.fc1 = nn.Linear(64*8*8, 128)  #input layer
        self.fc2= nn.Linear(128, 10) #hidden layer 이면서 output layer
    
    
    def forward(self,x):    #모듈에 있는 forward를 재정의 
        # self.covn1 : 배치, 채널, 높이, 너비 (x.shape)
        x = self.pool(F.relu(self.conv1(x)))   # 최초(64,3,32,32) --> conv1 (64,32,32,32) --> pool (64,32,16,16)
        x = self.pool(F.relu(self.conv2(x))) # -->conv2 (64,64,16, 16) --> pool (64, 64, 8,8) --> self.fc1에 넣기

        # 중요팁: 배치 크기를 64로 고정해서 x.view(64, -1) 하는 건 좋지 않다.
        # 배치 크기가 바뀌면 에러가 남. 대신 x = x.view(x.size(0), -1) 로 쓰자.
        x = x.view(x.size(0), -1) # x = x.view(64, -1) flatten 
        x = F.relu(self.fc1(x))   # 오류: 입력 행렬 크기 (64×1568) 와 가중치 행렬 (3136×128) 의 크기가 안 맞아서 곱셈 불가능 --> x.view(x.size(0), -1)
        out = self.fc2(x)
        return out
model = CNN()
model
device = torch.device('cuda' if torch.cuda.is_available() else 'cpu')
model.to(device)

# 손실함수 / 다중클래스에 적합 / 내부에 [softmax + log 손실계산] 있음
# 모델이 얼마나 틀렸는지 수치로 계산
criterion = nn.CrossEntropyLoss() # binary?

# optimizer 경사하강법(SGD) 변형/ 기울기를 조절
# 가중치를 업데이트(학습)하는 방법을 정함 --> 모델이 틀린 방향을 반영해서 가중치를 수정하는 역할 (오차를 줄이도록)
optimizer = optim.Adam(model.parameters(), lr=0.001)

```

```python
# 학습 루프 : 상기와 동일
from tqdm import tqdm
num_epochs = 5
for epoch in range(num_epochs):
    model.train()      # 드랍아웃, 배치노말, .. 등의 기능성 레이어를 활성화
    running_loss = 0.0 # 손실율 계산
    for inputs, labels in tqdm(train_loader) : 
        inputs, labels = inputs.to(device), labels.to(device)
        optimizer.zero_grad()  # 그래디언트 초기화: 기울기 누적되면 안되므로..??
        outputs =model(inputs)   # forward 호출 --> 예측
        loss = criterion(outputs, labels)  # 손실계산
        loss.backward () #역전파 (기울기를 계산)
        optimizer.step() #파라미터(가중치)를 업데이트 (학습!)
        running_loss += loss.item()
    print(f'Epoch: {epoch+1}/{num_epochs}, Loss: {running_loss/len(train_loader):.4f}')
```

```python
# 평가 : 상기와 동일
model.eval() #평가모드
correct = 0
total = 0
with torch.no_grad():   # 기울기 계산 비활성화 (메모리 절약)
    for inputs, labels in test_loader:
        inputs, labels = inputs.to(device), labels.to(device)
        output = model(inputs)
        _, predicted = torch.max(output.data, 1) #최대 확률 클래스 예측
        total += labels.size(0)
        correct += (predicted == labels).sum().item()
print(f'test accuracy : {(correct / total):.4f}')
```

## Early Stopping

```python
#### Early stopping : 훈련을 너무 오래하지 않도록 제어하는 기능
# 모델이 validation set(검증용 데이터)에서 더 이상 성능이 좋아지지 않으면 학습을 조기에 멈추는 기법입니다.
# 왜 필요할까?
# Epoch를 너무 많이 돌리면 → Overfitting(과적합) 발생
# → 학습 데이터에는 점점 잘 맞지만, 새로운 데이터에는 성능이 떨어짐
# 불필요하게 학습 오래 돌리면 → 시간 낭비

# Early Stopping을 쓰면
# → 학습이 충분히 좋아졌을 때 멈추고,
# → 과적합도 예방하고,
# → 시간을 절약할 수 있어요.


class EarlyStopping :
    def __init__(self, patience=5, min_delta=0):  # patience=5 : 5 epoch 연속 개선이 없으면 학습 중단
        self.patience = patience        #개선이 없더라도 몇 번 (epoch 기다릴지)
        self.min_delta = min_delta      #개선 됐다고 인정할 최소한의 변화량
        self.counter = 0                #개선이 없을 때 증가시킬 카운터
        self.best_loss = float ('inf')  # 지금까지 최소(최고성능) val_loss 저장
    def __call__(self, val_loss) :  #객체를 함수처럼 호출할 때 실행
        if val_loss < self.best_loss - self.min_delta:
            self.best_loss = val_loss
            self.counter=0
        else:
            self.counter += 1
            if self.counter >= self.patience :   #patience는 epoch 수
                return True #조기종료
        return False  # 학습 계속
early_stopping = EarlyStopping()
```

#### 훈련/검증함수 (accuracy 포함)

```python
# 훈련/검증 함수 (accuracy 포함)
from tqdm import tqdm
def train_epoch (model, loader, criterion, optimizer, device):
    model.train() 
    running_loss = 0.0  #배치손실을 누적할 변수
    correct = 0         #맞춘 샘플 수 누적
    total = 0           #전체 샘플 수 누적
    iterator =tqdm(enumerate(loader), total=len(loader), desc='Train')
    for batch_idx, (inputs, labels) in iterator:
            # enumerate(loader) → (인덱스, loader에서 꺼낸 값) 형태 반환
            # 여기서 loader는 DataLoader라 하나의 배치마다 (inputs, labels) 튜플을 반환
            # 즉, enumerate + unpacking을 한 번에 한 거예요.
                # loader = [ ([img1, img2], [0,1]), ([img3,img4],[1,0]) ]
                # print(batch_idx) ---> 0
                # print(inputs, labels) ----> [img1, img2] [0, 1]
        
        inputs, labels = inputs.to(device), labels.to(device)
        optimizer.zero_grad()  # 그래디언트 초기화: 이전 배치의 기울기가 누적되지 않도록!
        outputs =model(inputs)   # forward 호출 --> 예측
        loss = criterion(outputs, labels)  # 손실계산
        loss.backward () #역전파 (파라미터(가중치)에 대해 기울기를 계산)
        optimizer.step() #파라미터(가중치)를 업데이트 (학습! / lr에 따라 조금씩 업데이트)
        running_loss += loss.item() # loss.item() 텐서에서 값만 추출



        _, predicted = torch.max(outputs.data, 1) 
                # outputs.data는 모델의 예측 결과 배치 단위 로짓(logits), 예를 들어 (batch_size, num_classes) 형태
                # 예: (128, 10) → 배치 128개, 클래스 10개 확률/점수
                # torch.max(input, dim) 은 주어진 차원(dim)에서 최대값과 인덱스를 반환합니다.
                # _ → 최대값 자체는 필요 없으니 버리는 용도
                # predicted → 예측된 클래스 인덱스 (가장 max일때 인덱스가 최종 정답인 클래스)

        total += labels.size(0)
        correct += (predicted == labels).sum().item()  
            # 2️⃣ correct += (predicted == labels).sum().item()
            # (predicted == labels) → 맞춘 샘플은 True, 틀린 건 False
            # .sum() → True는 1로 계산되므로, 맞춘 샘플 수 합계
            # .item() → 텐서를 **숫자(scalar)**로 변환
            # correct → 지금까지 맞춘 샘플 수 누적


        iterator.set_postfix({'loss':f'{loss.item():.3f}'}, {'acc': f'{correct/total:.3f}'})
            # set_postfix를 사용하면 진행바 옆에 실시간 정보(예: loss, accuracy 등)를 표시 가능

    avg_loss = running_loss/len(loader)
            # 한 에포크 동안의 평균 손실 계산
            # running_loss → 배치 단위 손실 합계
            # len(loader) → 배치 개수
                # batch 1: loss = 0.9 / batch 2: loss = 0.8 / batch 3: loss = 0.7 /.../ batch 10: loss = 0.6
                # running_loss = 0.9 + 0.8 + 0.7 + ... + 0.6 = 7.5
                # avg_loss = 7.5 / 10 = 0.75

    accuracy = correct / total
            # 한 에포크 동안의 전체 정확도 계산
            # 전체 맞춘 샘플 / 전체 샘플

    print(f'Epoch: {batch_idx+1}, Loss: {running_loss/len(train_loader):.4f}, accuracy: {accuracy}')

    return avg_loss, accuracy
```

```python
# 학습 루프
num_epochs = 20
best_val_loss = float('inf')
for epoch in range(num_epochs):
    train_loss, train_acc = train_epoch(cnn_cifar10_model, train_loader, criterion, optimizer, device)
    scheduler.step()  #스케줄러 업데이트

    # early stopping check
    if early_stopping(best_val_loss):
        print(f'Early Stopping epoch : {epoch+1}')
        break
```

##### <<<<<< “왜 클래스를 만들고, forward라는 함수를 꼭 써야 해?” >>>>>>

- 딥러닝은 단순한 수학식이 아니라 계층(layer)으로 쌓인 네트워크예요.
- 클래스 안에서 __init__()은 층(layer)을 설계하는 부분 --> “이 모델은 이렇게 생겼다 → __init__”
- forward()는 데이터가 그 층을 통과할 때 어떤 계산을 하는지를 정의하는 부분 --> “데이터가 이렇게 흘러간다 → forward”
- 딥러닝은 “학습할 수 있는 함수”이기 때문에, 함수(=모델 구조)를 직접 정의해야 PyTorch가 역전파(backpropagation)로 학습을 시킬 수 있어요.
