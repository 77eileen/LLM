```python
import shutil

# 모든 드라이브 확인
drives = ['C:\\', 'D:\\', 'E:\\', 'F:\\']

for drive in drives:
    try:
        total, used, free = shutil.disk_usage(drive)
        print(f"\n{drive}")
        print(f"  총 용량: {total // (2**30):>6} GB")
        print(f"  사용 중: {used // (2**30):>6} GB")
        print(f"  여유 공간: {free // (2**30):>6} GB")
        print(f"  여유율: {free/total*100:>5.1f}%")
    except:
        pass  # 드라이브가 없으면 건너뜀
```

```python
# 허깅페이스 캐시 확인

import os
from pathlib import Path

# Hugging Face 캐시 위치
cache_dir = Path.home() / ".cache" / "huggingface" / "hub"

if cache_dir.exists():
    total_size = 0
    for dirpath, dirnames, filenames in os.walk(cache_dir):
        for f in filenames:
            fp = os.path.join(dirpath, f)
            if os.path.exists(fp):
                total_size += os.path.getsize(fp)
    
    print(f"HuggingFace 캐시 크기: {total_size / (1024**3):.2f} GB")
    print(f"캐시 위치: {cache_dir}")
else:
    print("HuggingFace 캐시가 없습니다.")
```
